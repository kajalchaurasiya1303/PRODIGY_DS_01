**Importing the Libraries and Displaying the data**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
data = pd.read_csv("Population of India.csv")

# Display the dataset
print("Population of India Dataset:")
print(data)
```

**BAR CHART OF Top10 STATES**

```python
# ---------------- BAR CHART ----------------

# Select top 10 states by population
top10 = data.sort_values("Population[50]", ascending=False).head(10)

plt.figure(figsize=(10, 6))

plt.bar(top10["State/UT"], top10["Population[50]"])

plt.xlabel("State/UT")
plt.ylabel("Population[50]")
plt.title("Top 10 States/UTs by Population")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**HISTOGRAM OF POPULATION IN INDIA**

```python
# ---------------- HISTOGRAM ----------------

plt.figure(figsize=(8, 5))

plt.hist(data["Population[50]"], bins=8, edgecolor="black")

plt.xlabel("Population")
plt.ylabel("Number of States/UTs")
plt.title("Distribution of Population in India")

plt.tight_layout()
plt.show()
```

**MALE VS FEMALE POPULATION IN INDIA**

```python
# Calculate total male and female population
male = data["Male"].sum()
female = data["Female"].sum()

# Create bar chart
plt.bar(["Male", "Female"], [male, female])

# Add labels
plt.xlabel("Gender")
plt.ylabel("Population")
plt.title("Male vs Female Population in India")

# Display chart
plt.show()
```

**MALE VS FEMALE POPULATION IN EACH STATES OF INDIA**

```python
# Set bar width
width = 0.4

# Create positions
x = range(len(data))

# Create bars
plt.figure(figsize=(14, 6))

plt.bar([i - width/2 for i in x], data["Male"], width, label="Male")
plt.bar([i + width/2 for i in x], data["Female"], width, label="Female")

# Labels
plt.xlabel("State/UT")
plt.ylabel("Population")
plt.title("Male vs Female Population in Each State/UT")

plt.xticks(x, data["State/UT"], rotation=90)

plt.legend()
plt.tight_layout()
plt.show()
```
